```python
import numpy as np
import pandas as pd

cancer = pd.read_csv("/content/drive/MyDrive/ovariantotal.csv") #stores all data in a table format inside the variable cancer
                                                                #reading the csv file automatically puts it into data frames
cancer.head()  #syntax: <data_frame_variable>.head/tail(<no. of data points>) head gives first 5 rows & viceversa
#cancer.info()   -# gives list all col types
```

```python
#in cancer data there is whole X feature and Y target we seprate them into X & Y matrix
#using drop function on the df will remove the column and store in x
#then we access the colunm and store it in y

X = cancer.drop("TYPE",axis=1) #syntax <df>.drop("<col name>",axis=0/1 {0for droping row else 1 for col})
Y = cancer["TYPE"]
```

```python
#we normalize every value in data set col wise to remove shear dominance of few features
#we intended to use minmax scalling method
from sklearn import preprocessing
min_max_scaler = preprocessing.MinMaxScaler() #MinMaxScaler is the method under module preprocessing if we ran directly on the data we can't store min&max values
X_scaled = min_max_scaler.fit_transform(X)
print(X_scaled)
```

```python
#now we split the data into training and testing
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X_scaled,Y,test_size=0.20,random_state=0)
```

```python
from tensorflow import keras
from keras.models import Sequential
#we need sequential model thats goes input->hiddenlayer->output hence from models inside keras we select sequential
from keras.layers import Dense #this is because we are having hidden layer which is a dence NN

model = Sequential()
model.add(Dense(32,activation="relu",input_dim=49))
model.add(Dense(32,activation="relu"))
model.add(Dense(1,activation="sigmoid"))

```

```python
#now we compile it
model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
model.fit(X_train,Y_train,batch_size=32,epochs=15)
```

```python
Y_hat = model.predict(X_test)
print(Y_hat)
```

```python
Y_hat = np.where(Y_hat>0.5,1,0)
Y_hat = Y_hat.astype(int)
np.column_stack((Y_hat,Y_test))
```

```python
#knowing the accuracy
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
print(confusion_matrix(Y_hat,Y_test))
print(accuracy_score(Y_hat,Y_test))
```
