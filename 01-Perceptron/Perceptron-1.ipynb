# First perceptron model
so in this we will take one single input 1,2 then give some weights and try to get the required output
hence here, batch = 1 batch_size=1, epoch=3



```python
import numpy as np
```

```python
x = np.array([1.0,2.0])             #inputs
y = 1                               #target
weights = np.array([0.2,-0.5])      #weights
bias = 0.1
eta = 0.2                           #learning rate
```

```python
for epoch in range(1,4):
  print(f"\nepoch:{epoch}")
  net_input = np.dot(weights,x)+bias
  y_hat = 1 if net_input>0 else 0
  error = y-y_hat
  print(f"net_input:{net_input:.2f}")
  print(f"predicted y:{y_hat}")
  print(f"loss:{error}")

  #updating
  weights = weights + eta*error*x
  bias = bias+eta*error
  print(f"updated weights:{weights}")
  print(f"updated bias:{bias:.2f}")
```
