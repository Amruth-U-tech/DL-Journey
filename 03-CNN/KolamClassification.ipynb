**ConvNext tiny model**


```python
from tensorflow.keras.applications import ConvNeXtTiny

from tensorflow.keras.layers import Dense, Dropout, GlobalAveragePooling2D, BatchNormalization
from tensorflow.keras.models import Sequential
from tensorflow.keras.optimizers import Adam

convnexttiny = ConvNeXtTiny(weights="imagenet",
                            include_top=False,
                            input_shape = (224,224,3))
convnexttiny.trainable = False

modelCT = Sequential()
modelCT.add(convnexttiny)

modelCT.add(GlobalAveragePooling2D())

modelCT.add(Dense(256,activation="relu"))
modelCT.add(BatchNormalization())
modelCT.add(Dropout(0.4))

modelCT.add(Dense(256,activation="relu"))
modelCT.add(BatchNormalization())
modelCT.add(Dropout(0.5))

modelCT.add(Dense(12,activation="softmax"))

modelCT.compile(optimizer=Adam(learning_rate=3e-5),
                loss="categorical_crossentropy",
                metrics=["accuracy"])
modelCT.summary()
```
