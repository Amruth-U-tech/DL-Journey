```python
#we first import the packages required to do data preprocessing
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import warnings

warnings.filterwarnings("ignore")  #this is to supress any warnings we get

train_dir = "/content/drive/MyDrive/alzheimer dataset/train"
test_dir = "/content/drive/MyDrive/alzheimer dataset/test"
```

```python
#now we craerte an image generator object which we take the image path and get through the agumentation
#and preprocessing

train_datagen = ImageDataGenerator(rescale=1./255,
                                   shear_range=0.4,
                                   zoom_range=0.2,
                                   width_shift_range=0.2,
                                   height_shift_range=0.2,
                                   rotation_range=90,
                                   horizontal_flip=True,
                                   fill_mode="nearest"
                                   )
test_datagen = ImageDataGenerator(rescale=1./255)

#loading image and lables, ensuring grayscale

train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(128,128),            #resizing the image to 128x128
    color_mode="grayscale",
    class_mode="categorical"
)

test_generator = test_datagen.flow_from_directory(test_dir,target_size=(128,128),color_mode="grayscale",class_mode="categorical")

```

```python
from keras.models import Sequential    #cause cnn are sequential
from keras.layers import Conv2D         #as we are dealing with 2d data set
from keras.layers import Dense          #each node connected to each and evry other node
from keras.layers import MaxPooling2D   #pooling layer to extract imp faetures
from keras.layers import Flatten        #then to send it to fully connected neural network
from keras.layers import Dropout
import numpy as np

#u can import dropout too
```

```python
#define the cnn model
CNN1 = Sequential()

#build cnn layers one by one
CNN1.add(Conv2D(256,(3,3),activation="relu",input_shape=(128,128,1)))
CNN1.add(MaxPooling2D(pool_size=(2,2)))

CNN1.add(Conv2D(64,(3,3),activation="relu"))
CNN1.add(MaxPooling2D(pool_size=(2,2)))

CNN1.add(Conv2D(256,(3,3),activation="relu"))
CNN1.add(MaxPooling2D(pool_size=(2,2)))

CNN1.add(Conv2D(128,(3,3),activation="relu"))
CNN1.add(MaxPooling2D(pool_size=(2,2)))

CNN1.add(Conv2D(256,(3,3),activation="relu"))
CNN1.add(MaxPooling2D(pool_size=(2,2)))

#now we need flatten
CNN1.add(Flatten())

#then we pass that to fully connected neural network
CNN1.add(Dense(128,activation="relu"))
CNN1.add(Dropout(0.2))
CNN1.add(Dense(64,activation="relu"))

#output layer cause this is multi class classification we will have softmax as activation function
CNN1.add(Dense(4,activation="softmax")) #ther are 4 classes so 4 nodes

#now for backpropogation and everything we use the compile
CNN1.compile(optimizer="adam",loss="categorical_crossentropy",metrics=["accuracy"])

CNN1.summary()  #summary gives the summary and claculations of our model completly

#here we just built the model and not yet passed the data untill and unless we use fit
```

```python
#now we can train it
CNN1.fit(train_generator,epochs=40)  #when batch size is not defined anywhere it should have been ther the default batchsize=32
```

```python
loss,accuracy1 = CNN1.evaluate(test_generator)
#running on the test data set

#printing the test accuracy
print(f"the accuracy of the model pridiction is:{accuracy1}")
```
